```python
from typing import Literal
from pydantic import BaseModel, Field
from langchain_core.documents import Document
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.tools import tool
from langchain_groq import ChatGroq
from langgraph.func import entrypoint, task
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.store.memory import InMemoryStore
from langgraph.types import interrupt, Command, RetryPolicy
from langgraph.utils.config import get_store
from langgraph.errors import GraphInterrupt # استيراد دالة التوقف هنا

# 1. RAG Setup
raw_docs = [
    Document(page_content="King Saud Stadium costs 500 SAR per hour. Cancellation requires 24 hours notice."),
    Document(page_content="Al-Awwal Park costs 800 SAR per hour. Open from 4 PM to 12 AM.")
]
text_splitter = RecursiveCharacterTextSplitter(chunk_size=50, chunk_overlap=10)
chunks = text_splitter.split_documents(raw_docs)
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
vectorstore = FAISS.from_documents(chunks, embeddings)
retriever = vectorstore.as_retriever()

# 2. Tools Definition
@tool
def book_stadium_tool(stadium_name: str, time: str) -> str:
    """Book a soccer stadium for a specific time."""
    return f"Confirmed booking for {stadium_name} at {time}."

# 3. Router Definition
class RouteDecision(BaseModel):
    next_step: Literal["rag_search", "book_stadium", "escalation"] = Field(
        description="Decide next step based on query."
    )

# استخدمنا نموذج Qwen الذي اخترته أنت
llm = ChatGroq(model="qwen/qwen3.6-27b", temperature=0)
router_llm = llm.with_structured_output(RouteDecision)
tools_llm = llm.bind_tools([book_stadium_tool])

# 4. Tasks (Functional API)
@task(retry=RetryPolicy(max_attempts=3))
def route_query(query: str) -> RouteDecision:
    return router_llm.invoke(query)

@task
def search_knowledge(query: str) -> str:
    results = retriever.invoke(query)
    info = " ".join([d.page_content for d in results])
    return llm.invoke(f"Answer based on info: {info}\nQuery: {query}").content

@task
def process_booking(query: str) -> str:
    store = get_store()
    store.put("user_data", "last_booking_status", {"status": "pending_approval"})
    approval = interrupt("System Paused: Approve this booking? (yes/no)")
    if approval.lower() == "yes":
        return f"Approved. {tools_llm.invoke(query).tool_calls}"
    return "Booking rejected."

@task
def fallback_task(query: str) -> str:
    return "Escalating to human support."

my_store = InMemoryStore()
my_checkpointer = InMemorySaver()

# 5. Entrypoint
@entrypoint(checkpointer=my_checkpointer, store=my_store)
def stadium_agent(query: str) -> str:
    try:
        decision = route_query(query).result()
        if decision.next_step == "rag_search":
            return search_knowledge(query).result()
        elif decision.next_step == "book_stadium":
            return process_booking(query).result()
        else:
            return fallback_task(query).result()
    except GraphInterrupt:
        # السماح لـ LangGraph بإيقاف النظام لانتظار الموافقة البشرية
        raise
    except Exception as e:
        return f"Error occurred: {str(e)}"
```

```python
thread_config = {"configurable": {"thread_id": "session_1"}}

print("--- Test 1: RAG Search ---")
print(stadium_agent.invoke("Cost of King Saud Stadium?", thread_config))

print("\n--- Test 2: HITL Pause ---")
print(stadium_agent.invoke("Book King Saud Stadium for 9 PM", thread_config))

print("\n--- Test 3: HITL Resume ---")
print(stadium_agent.invoke(Command(resume="yes"), thread_config))

print("\n--- Test 4: Cross-Thread Long-Term Memory ---")
new_thread = {"configurable": {"thread_id": "session_99"}}
# نستدعي الذاكرة التي عرفناها برمجياً لقراءتها بدون أخطاء
memory = my_store.get("user_data", "last_booking_status")
if memory:
    print(f"Memory retrieved from entirely new thread: {memory.value}")
else:
    print("Memory not found.")
```
